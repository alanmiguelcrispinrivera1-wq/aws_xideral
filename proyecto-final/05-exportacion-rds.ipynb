# 05 - Exportar KPIs de Gold a RDS (MySQL)

````python
import os
import getpass
import json
import boto3
import pandas as pd
from sqlalchemy import create_engine, text

BUCKET = "xideralaws-curso-proyecto-alan"

mysql_host = os.getenv("MYSQL_HOST") or input("Endpoint de RDS MySQL: ")
mysql_port = int(os.getenv("MYSQL_PORT", "3306"))
mysql_user = os.getenv("MYSQL_USER") or input("Usuario de MySQL: ")
mysql_db = os.getenv("MYSQL_DB") or input("Base de datos (schema) donde tienes permiso de crear tablas: ")
mysql_password = os.getenv("MYSQL_PASSWORD") or getpass.getpass("Contraseña de MySQL: ")

connection_string = f"mysql+pymysql://{mysql_user}:{mysql_password}@{mysql_host}:{mysql_port}/{mysql_db}"
engine = create_engine(connection_string)

with engine.connect() as conn:
    conn.execute(text("SELECT 1"))

print("Conexion a RDS exitosa")
print("Host:", mysql_host)
print("Base de datos:", mysql_db)
print("Usuario:", mysql_user)
````

### Función auxiliar para leer resultados de `gold_data/`
Todos los resultados (excepto la correlación, que guardamos como JSON, y `trip_duration`, que ahora se calcula aparte) son Parquet pequeño, por lo tanto, pandas los puede leer directamente desde S3 sin necesidad de utilizar Spark.

````python
def leer_gold_parquet(nombre_carpeta):
    ruta = f"s3://{BUCKET}/gold_data/{nombre_carpeta}/"
    df = pd.read_parquet(ruta)
    print(f"{nombre_carpeta}: {len(df)} filas leidas")
    return df

s3_client = boto3.client("s3", region_name="us-west-1")
````

### Exportar las tablas Parquet a RDS
Usamos `if_exists="replace"` a propósito: estos KPIs se recalculan por completo cada vez que corre el pipeline (no se acumulan), así que reemplazar la tabla entera es la forma correcta de mantenerlos sincronizados con la última corrida. Esto la vuelve idempotente gracias al diseño que estamos trabajando, igual que el resto del pipeline.

`trip_duration` ya no está en esta lista -- se exporta en la celda siguiente, calculado directo desde Silver en vez de leído desde Gold.

````python
tablas_parquet = [
    "trips_by_day",
    "peak_hours",
    "weekday_weekend",
    "fare_and_tip",
    "geographic_demand",
]

for nombre in tablas_parquet:
    df = leer_gold_parquet(nombre)
    nombre_tabla = f"gold_{nombre}"
    df.to_sql(nombre_tabla, engine, if_exists="replace", index=False)
    print(f"  -> exportado a tabla RDS: {nombre_tabla}")
````

### Cálculo directo desde Silver (sin Spark)
Calculamos la mediana de duración directo desde `clean_data/` (Silver) con pyarrow -- es la misma columna `duration_minutes` que ya limpiamos en el 03.

Usamos la **mediana** (no el promedio) para que coincida con la metodología del resto del proyecto y con lo que espera el dashboard de Streamlit (`duracion_mediana_min`): la mediana es menos sensible a viajes atípicos que el promedio.

````python
import gc
import pyarrow.parquet as pq
import pyarrow.fs as pafs

s3_pa = pafs.S3FileSystem(region="us-west-1")

# Tope de filas que tomamos POR ARCHIVO -- asi el pico de memoria queda
# acotado sin importar que tan grande sea el tipo de taxi completo (fhvhv
# por si solo pesa 10 GiB en Silver, cargar esa columna completa de golpe
# revienta la RAM del t2.medium). Leemos un archivo a la vez y muestreamos,
# el mismo principio que "percentile_approx" de Spark: mediana aproximada
# en vez de exacta sobre todos los datos.
MUESTRA_MAXIMA_POR_ARCHIVO = 20000

filas_duracion = []
for tipo in ["yellow", "green", "fhv", "fhvhv"]:
    prefijo = f"{BUCKET}/clean_data/{tipo}/"
    archivos = [
        info.path
        for info in s3_pa.get_file_info(pafs.FileSelector(prefijo, recursive=True))
        if info.path.endswith(".parquet")
    ]

    muestras = []
    for ruta_archivo in archivos:
        tabla = pq.read_table(ruta_archivo, columns=["duration_minutes"], filesystem=s3_pa)
        columna = tabla.column("duration_minutes").to_pandas()
        if len(columna) > MUESTRA_MAXIMA_POR_ARCHIVO:
            columna = columna.sample(MUESTRA_MAXIMA_POR_ARCHIVO, random_state=42)
        muestras.append(columna)
        del tabla, columna
    gc.collect()

    serie_muestra = pd.concat(muestras, ignore_index=True)
    filas_duracion.append({
        "taxi_type": tipo,
        "duracion_mediana_min": round(serie_muestra.median(), 2),
    })
    print(f"{tipo}: mediana calculada sobre muestra de {len(serie_muestra):,} viajes (de {len(archivos)} archivos)")
    del muestras, serie_muestra
    gc.collect()

df_trip_duration = pd.DataFrame(filas_duracion)
df_trip_duration.to_sql("gold_trip_duration", engine, if_exists="replace", index=False)
print(f"gold_trip_duration: {len(df_trip_duration)} filas exportadas (mediana aproximada, sin Spark)")
print(df_trip_duration)
````

### Exportar la matriz de correlación
Esta la guardamos como JSON en el notebook 04, no como Parquet. La convertimos a formato "largo" (una fila por par de tipos) para que sea fácil de consultar con SQL normal.

````python
respuesta = s3_client.get_object(Bucket=BUCKET, Key="gold_data/cross_taxi_correlation/correlation_matrix.json")
matriz = json.loads(respuesta["Body"].read())

filas = []
for tipo_a, correlaciones in matriz.items():
    for tipo_b, valor in correlaciones.items():
        filas.append({"taxi_type_a": tipo_a, "taxi_type_b": tipo_b, "correlacion": valor})

df_correlacion = pd.DataFrame(filas)
df_correlacion.to_sql("gold_cross_taxi_correlation", engine, if_exists="replace", index=False)
print(f"gold_cross_taxi_correlation: {len(df_correlacion)} filas exportadas")
````

### Verificación
Confirma en RDS (no en S3) que cada tabla tiene filas, así nos aseguramos de que no solo se creó la tabla, sino que sí llegamos a los datos.

````python
nombres_tablas = [f"gold_{n}" for n in tablas_parquet] + ["gold_trip_duration", "gold_cross_taxi_correlation"]

with engine.connect() as conn:
    for tabla in nombres_tablas:
        resultado = conn.execute(text(f"SELECT COUNT(*) FROM {tabla}")).scalar()
        print(f"{tabla}: {resultado} filas en RDS")
````

````python
engine.dispose()
print("Conexion a RDS cerrada")
````
